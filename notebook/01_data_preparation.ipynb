```python
def normalize_name(name):
    return (
        str(name)
        .strip()
        .lower()
        .replace(" ", "")
        .replace("_", "")
    )

documentation_names = {
    normalize_name(name): name
    for name in features_df["Name"]
}

dataset_names = {
    normalize_name(name): name
    for name in train_clean.columns
}

documentation_not_in_dataset = [
    documentation_names[name]
    for name in documentation_names
    if name not in dataset_names
]

dataset_not_in_documentation = [
    dataset_names[name]
    for name in dataset_names
    if name not in documentation_names
]

print("Documentation fields without normalized dataset match:")
print(documentation_not_in_dataset)

print("\nDataset columns without normalized documentation match:")
print(dataset_not_in_documentation)
```

## Final Data Preparation Summary

The UNSW-NB15 training and testing datasets were systematically
audited before preparation.

### Data quality findings

- No missing values were found.
- No duplicate rows were found.
- No infinite numerical values were found.
- No negative numerical values were found.
- Training and testing datasets contain the same 45 columns.
- Training and testing columns have matching data types.
- `id` contains a unique value for every record and was treated as
  a record identifier rather than a predictive feature.
- Categorical variables were inspected for their distributions and
  train/test category differences.
- Some categories occur only in one dataset split, particularly
  within `proto` and `state`. These categories were retained rather
  than removed.
- The `is_ftp_login` column is documented as binary, but the prepared
  dataset contains values 0, 1, 2, and 4. The values 2 and 4 occur
  only in FTP records and correspond exactly to `ct_ftp_cmd`. These
  values were retained because the supplied documentation does not
  establish a justified recoding rule.

### Preparation decisions

No unnecessary rows or observations were removed. Since the dataset
contained no missing, duplicate, infinite, or negative values, no
imputation or deletion was required.

Column names were standardized by removing leading and trailing
whitespace.

The supplied feature documentation contains 49 documented fields,
while the prepared train/test datasets contain 45 columns. Naming
differences were documented rather than arbitrarily changing the
dataset schema. Fields not available in the prepared datasets were
not reconstructed or imputed.

The resulting datasets preserve the available source information
while documenting the identified schema and data-definition
differences.

## Assumptions and Limitations

1. The prepared UNSW-NB15 train/test CSV files were treated as the
   source data for this analysis.

2. The supplied feature-description file was used as supporting
   documentation for interpreting the dataset columns.

3. Differences between documentation names and dataset column names
   were treated as schema/naming differences unless there was
   sufficient evidence to identify them as data-quality errors.

4. Rare categorical values were retained because rarity alone does
   not establish that a value is invalid.

5. Extreme numerical values were retained because their magnitude
   alone does not establish that they are erroneous network
   measurements.

6. The `id` column was retained for record identification but should
   not be used as a predictive feature in subsequent machine-learning
   models.

7. The `is_ftp_login` values 2 and 4 were retained because the
   available documentation does not provide enough information to
   justify converting them to another value.
